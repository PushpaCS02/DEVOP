Complete EDA workflow – business questions

```python
#Key business totals
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = df["Order ID"].nunique()
total_quantity = df["Quantity"].sum()

print("Total sales:", round(total_sales, 2))
print("Total profit:", round(total_profit, 2))
print("Unique orders:", total_orders)
print("Units sold:", total_quantity)
```

```python
#Sales and profit by region
region_summary = df.groupby("Region")[["Sales", "Profit"]].sum().sort_values("Sales", ascending=False)
region_summary
region_summary.plot(kind="bar", figsize=(9, 5))
plt.title("Sales and Profit by Region")
plt.xlabel("Region")
plt.ylabel("Amount")
plt.xticks(rotation=0)
plt.show()
```

```python
# Top 10 products by sales
top_products = (df.groupby("Product Name")["Sales"]
                .sum().sort_values(ascending=False).head(10))
top_products
plt.figure(figsize=(10, 6))
top_products.sort_values().plot(kind="barh")
plt.title("Top 10 Products by Sales")
plt.xlabel("Total sales")
plt.ylabel("Product")
plt.show()
```

```python
#Profit by sub-category
subcategory_profit = df.groupby("Sub-Category")["Profit"].sum().sort_values()
subcategory_profit
plt.figure(figsize=(10, 7))
subcategory_profit.plot(kind="barh")
plt.title("Profit by Sub-Category")
plt.xlabel("Total profit")
plt.ylabel("Sub-category")
plt.show()
```

```python
#Monthly sales and profit trend
monthly_performance = df.groupby("Order Month")[["Sales", "Profit"]].sum().sort_index()
monthly_performance.head()monthly_performance.plot(figsize=(12, 5), marker="o")
plt.title("Monthly Sales and Profit")
plt.xlabel("Order month")
plt.ylabel("Amount")
plt.xticks(rotation=75)
plt.tight_layout()
plt.show()
```

```python
#Segment analysis
segment_summary = df.groupby("Segment").agg(
    Orders=("Order ID", "nunique"),
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum")
).sort_values("Sales", ascending=False)
segment_summary
segment_summary["Sales"].plot(kind="bar", figsize=(8, 5))
plt.title("Sales by Customer Segment")
plt.xlabel("Customer segment")
plt.ylabel("Total sales")
plt.xticks(rotation=0)
plt.show()
```
